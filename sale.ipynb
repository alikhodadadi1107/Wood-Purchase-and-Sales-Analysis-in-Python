```python
import jdatetime as jdt
import jalali_pandas
```

```python
def to_jalali_date(column, year_dir='L'): 
    if column.dtype == 'object':
        if '/' in column[0]:
            mod_col = [x.split('/') for x in column]
        elif '-' in column[0]:
            mod_col = [x.split('-') for x in column]
        elif '.' in column[0]:
            mod_col = [x.split('.') for x in column]
    
    elif column.dtype == 'int64':
        str_col = column.astype('str')
        if len(str(column[0])) == 8:
            if year_dir == 'L':
                mod_col = [[x[0:4], x[4:6], x[6:]] for x in str_col]
            elif year_dir == 'R':
                mod_col = [[x[0:2], x[2:4], x[4:]] for x in str_col]
        elif len(str(column[0])) == 6:
            mod_col = [[x[0:2], x[2:4], x[4:]] for x in str_col]

    if year_dir == 'L':
        year = [i[0] for i in mod_col]
        month = [i[1] for i in mod_col]
        day = [i[2] for i in mod_col]
    else:
        year = [i[2] for i in mod_col]
        month = [i[1] for i in mod_col]
        day = [i[0] for i in mod_col]

    if len(year[0]) == 2:
        if int(year[0]) < 50:
            year = ['14'+j for j in year]
        elif int(year[0]) > 50:
            year = ['13'+j for j in year]

    new_correct_date = []
    for n in range(len(year)):
        y = int(year[n])
        m = int(month[n])
        d = int(day[n])
        new_correct_date.append(jdt.date(y, m, d))  # jdt.datetime
    return new_correct_date
```

```python
wood1['Inv_Date'] = to_jalali_date(wood1['date'])
wood1.sample(5)
```

```python
wood1['year']    = wood1.Inv_Date.jalali.year
wood1['month']   = wood1.Inv_Date.jalali.month
wood1['day'] = wood1.Inv_Date.jalali.day

wood1 .sample(5)
```

```python
best_customer = wood1.groupby(
                            ['Customer_Name' , 'year' , 'month']
                        
                            )['Net Amount'].sum().reset_index()

best_customer.head(12)
```

```python
wood1.groupby(['Customer_Name' , 'year' , 'month'] ) .ngroups
```

```python
top_customer_month =best_customer.loc[best_customer.groupby( ['Customer_Name' , 'year' , 'month'])['Net Amount'].idxmax()]
top_customer_month
```

```python
sort = top_customer_month.sort_values( by= ['month' , 'Net Amount'] , ascending= [True , False])
sort.reset_index()
```

```python
top_customer_1_month = top_customer_month.groupby(['year', 'month']).first().reset_index()
top_customer_1_month
```
